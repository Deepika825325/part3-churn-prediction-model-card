## Leakage Prevention Strategy

Customer churn prediction must only use information that would have been available at the snapshot date.

To prevent data leakage:

- Only pre-snapshot customer information is used.
- Future purchases are excluded.
- Future support tickets are excluded.
- Future campaign outcomes are excluded.
- Churn labels are used only as the target variable.

The modeling snapshot provided in the dataset is assumed to contain features generated before the churn observation window.

```python
# Inspect Snapshot Dataset
snapshot.columns.tolist()
```

```python
# create Leakage Audit Table
leakage_audit = pd.DataFrame({
    "Feature Source": [
        "Customer Profile",
        "Historical Orders",
        "Historical Support Tickets",
        "Historical Web Activity",
        "Historical Campaign Data",
        "Churn Label"
    ],
    
    "Used in Features": [
        "Yes",
        "Yes",
        "Yes",
        "Yes",
        "Yes",
        "No"
    ],
    
    "Reason": [
        "Available before snapshot",
        "Generated before snapshot",
        "Generated before snapshot",
        "Generated before snapshot",
        "Generated before snapshot",
        "Target variable only"
    ]
})

leakage_audit
```

```python
# Check for Suspicious Columns
for col in snapshot.columns:
    print(col)
```

```python
# Create a leakage Screening cell
suspicious_keywords = [
    "future",
    "next",
    "post",
    "churn",
    "target"
]

potential_leakage = []

for col in snapshot.columns:
    
    for keyword in suspicious_keywords:
        
        if keyword.lower() in col.lower():
            
            potential_leakage.append(col)

potential_leakage
```

### Leakage Assessment Findings

The modeling snapshot contains one target variable:

- churn_next_60d

This column represents whether the customer churned within the future 60-day prediction window and therefore cannot be used as an input feature.

The following columns were also excluded from modeling:

- customer_id (identifier only)
- snapshot_date (administrative field)
- split (predefined data partition indicator)

All remaining variables represent customer behavior observed before the snapshot date and are therefore considered valid predictors.

# 8. Train/Test Split

The dataset includes a predefined split column. This split is used to ensure consistency and reproducibility during model evaluation.

```python
# 8.1 - Inspect split Distribution
snapshot["split"].value_counts()
```

```python
# 8.2- Split summary table
split_summary = pd.DataFrame({
    "Count": snapshot["split"].value_counts(),
    "Percentage": (
        snapshot["split"].value_counts(normalize=True)*100
    ).round(2)
})

split_summary
```

```python
# 8.3- Create Train/validation/Test Datasets
train_df = snapshot[
    snapshot["split"] == "train"
]

val_df = snapshot[
    snapshot["split"] == "validation"
]

test_df = snapshot[
    snapshot["split"] == "test"
]
```

```python
# 8.4 - Create X and Y for each split
feature_cols = [
    col for col in snapshot.columns
    if col not in [
        "customer_id",
        "snapshot_date",
        "split",
        "churn_next_60d"
    ]
]

target = "churn_next_60d"
```

```python
X_train = train_df[feature_cols]
y_train = train_df[target]

X_val = val_df[feature_cols]
y_val = val_df[target]

X_test = test_df[feature_cols]
y_test = test_df[target]
```

```python
# 8.7 - verify Shape
print("Training Set :", X_train.shape)
print("Validation Set :", X_val.shape)
print("Test Set :", X_test.shape)
```

```python
# 8.8 - Verify Target Distribution Across splits
print("Train")
print(y_train.value_counts(normalize=True))

print("\nValidation")
print(y_val.value_counts(normalize=True))

print("\nTest")
print(y_test.value_counts(normalize=True))
```

```python
# 8.9 - Split Summary Table
split_details = pd.DataFrame({
    "Dataset": [
        "Train",
        "Validation",
        "Test"
    ],
    
    "Rows": [
        X_train.shape[0],
        X_val.shape[0],
        X_test.shape[0]
    ],
    
    "Features": [
        X_train.shape[1],
        X_val.shape[1],
        X_test.shape[1]
    ]
})

split_details
```
