## Leakage Prevention Strategy

Customer churn prediction must only use information that would have been available at the snapshot date.

To prevent data leakage:

- Only pre-snapshot customer information is used.
- Future purchases are excluded.
- Future support tickets are excluded.
- Future campaign outcomes are excluded.
- Churn labels are used only as the target variable.

The modeling snapshot provided in the dataset is assumed to contain features generated before the churn observation window.

### Outlier Assessment

An IQR-based analysis was performed to identify potential outliers across numerical features.

The objective was to determine whether extreme values represent data quality issues or genuine customer behavior.

```python
# Numerical Features

numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# IQR-Based Outlier Assessment

outlier_summary = []

for col in numerical_features:

    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outlier_count = (
        (
            X[col] < lower_bound
        )
        |
        (
            X[col] > upper_bound
        )
    ).sum()

    outlier_summary.append({
        "Feature": col,
        "Outlier Count": outlier_count,
        "Outlier %": round(
            outlier_count / len(X) * 100,
            2
        )
    })

outlier_summary = pd.DataFrame(
    outlier_summary
)

outlier_summary.sort_values(
    by="Outlier Count",
    ascending=False
)
```

### Outlier Treatment Decision

No outlier capping or removal was performed.

Outliers were primarily observed in support-related and behavioral features such as Ticket Count, Resolution Time, Negative Ticket Rate, and Return Rate.

These observations likely represent genuine customer behavior rather than data quality issues and may contain valuable churn signals. Therefore, all outliers were retained for modeling.

# 9. Preprocessing Pipeline

Before model training, the data is preprocessed to handle missing values and transform features into a machine-learning-ready format.

The preprocessing pipeline includes:

- Missing value imputation
- Numerical feature scaling
- Categorical feature encoding
- Leakage-safe transformations using Scikit-Learn Pipelines

```python
# 9.1 - Verify Feature Types
print("Categorical Features:", len(categorical_features))
print("Numerical Features:", len(numerical_features))
```

```python
# 9.2 - Display Features lists
print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)
```

### Missing Value Handling

Only the `loyalty_tier` feature contains missing values.

Missing categorical values will be imputed using the most frequent category.

```python
X_train[categorical_features].isnull().sum()
```

```python
# 9.3 - Create Numerical Pipeline
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)
```

```python
# 9.4 - Create Categorical Pipeline
from sklearn.preprocessing import OneHotEncoder
categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)
```

```python
# Column-wise Transformations
from sklearn.compose import ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numerical_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)
```

```python
preprocessor
```

```python
X_train_processed = preprocessor.fit_transform(
    X_train
)

X_val_processed = preprocessor.transform(
    X_val
)

X_test_processed = preprocessor.transform(
    X_test
)
```

```python
print("Train Shape:", X_train_processed.shape)
print("Validation Shape:", X_val_processed.shape)
print("Test Shape:", X_test_processed.shape)
```

### Leakage Prevention

The preprocessing pipeline was fitted only on the training dataset.

The validation and test datasets were transformed using the fitted pipeline, ensuring that no information from future datasets influenced model training.

### Key Findings

- The dataset contains 6 categorical and 19 numerical features.
- `loyalty_tier` is the only feature with missing values and was handled through imputation.
- Categorical features were encoded using OneHotEncoder, increasing the feature count from 25 to 43.
- All preprocessing steps were applied through a leakage-safe pipeline fitted only on the training data.

# 10. Baseline Model

Logistic Regression is used as the baseline model.

It provides a simple and interpretable benchmark against which more advanced models can be compared.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

lr_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            LogisticRegression(
                random_state=42,
                max_iter=1000
            )
        )
    ]
)
```

```python
# Train Model
lr_pipeline.fit(
    X_train,
    y_train
)
```

```python
# Validation Predictions
```

```python
y_val_pred_lr = lr_pipeline.predict(
    X_val
)

y_val_prob_lr = lr_pipeline.predict_proba(
    X_val
)[:,1]
```

```python
# Evaluate Baseline Model
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)
```

```python
lr_results = {

    "Accuracy":
    accuracy_score(
        y_val,
        y_val_pred_lr
    ),

    "Precision":
    precision_score(
        y_val,
        y_val_pred_lr
    ),

    "Recall":
    recall_score(
        y_val,
        y_val_pred_lr
    ),

    "F1":
    f1_score(
        y_val,
        y_val_pred_lr
    ),

    "ROC_AUC":
    roc_auc_score(
        y_val,
        y_val_prob_lr
    ),

    "PR_AUC":
    average_precision_score(
        y_val,
        y_val_prob_lr
    )
}

lr_results
```

```python
pd.DataFrame(
    lr_results,
    index=["Logistic Regression"]
).T
```

# 11. Advanced Model 1: Random Forest

Random Forest is an ensemble learning algorithm that combines multiple decision trees to improve predictive performance and reduce overfitting.

It is expected to capture more complex churn patterns than Logistic Regression.

# 12. Advanced Model 2: Gradient Boosting

Gradient Boosting builds decision trees sequentially, where each new tree attempts to correct errors made by previous trees.

This approach often achieves better predictive performance than bagging-based methods such as Random Forest.

```python
from sklearn.ensemble import GradientBoostingClassifier

gb_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            GradientBoostingClassifier(
                n_estimators=200,
                learning_rate=0.05,
                max_depth=3,
                random_state=42
            )
        )
    ]
)
```

# 18. Threshold Optimization

The default classification threshold of 0.50 may not be optimal for churn prediction.

Different thresholds were evaluated to identify the best balance between Precision and Recall from a business perspective.

```python
# Evaluate Multiple Thresholds
thresholds = np.arange(0.1, 0.91, 0.05)

threshold_results = []

for threshold in thresholds:

    predictions = (
        y_val_prob_lr >= threshold
    ).astype(int)

    threshold_results.append({

        "Threshold": threshold,

        "Precision":
        precision_score(
            y_val,
            predictions
        ),

        "Recall":
        recall_score(
            y_val,
            predictions
        ),

        "F1":
        f1_score(
            y_val,
            predictions
        )

    })

threshold_df = pd.DataFrame(
    threshold_results
)

threshold_df
```

```python
# Find Best F1 Threshold
best_threshold_row = threshold_df.loc[
    threshold_df["F1"].idxmax()
]

best_threshold_row
```

```python
# Plot threshold performance
plt.figure(figsize=(8,5))

plt.plot(
    threshold_df["Threshold"],
    threshold_df["Precision"],
    label="Precision"
)

plt.plot(
    threshold_df["Threshold"],
    threshold_df["Recall"],
    label="Recall"
)

plt.plot(
    threshold_df["Threshold"],
    threshold_df["F1"],
    label="F1 Score"
)

plt.xlabel("Threshold")
plt.ylabel("Score")

plt.title(
    "Threshold Optimization"
)

plt.legend()

plt.savefig(
    "outputs/threshold_optimization.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
threshold_df.to_csv(
    "outputs/threshold_analysis.csv",
    index=False
)
```

```python
selected_threshold = best_threshold_row["Threshold"]

print(
    f"Selected Threshold: {selected_threshold:.2f}"
)
```

### 18B. Cost-Based Threshold Evaluation

While F1-score was used as the primary optimization metric, a business-cost evaluation was also performed.

For this analysis:

- False Positive Cost = ₹500
- False Negative Cost = ₹5000

The objective is to evaluate whether the F1-optimal threshold also minimizes expected business cost.

```python
cost_results = []

for _, row in threshold_df.iterrows():

    threshold = row["Threshold"]

    preds = (
        y_val_prob_lr >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        preds
    ).ravel()

    total_cost = (
        fp * 500
        +
        fn * 5000
    )

    cost_results.append({
        "Threshold": threshold,
        "FP": fp,
        "FN": fn,
        "Business Cost": total_cost
    })

cost_df = pd.DataFrame(cost_results)

cost_df.sort_values(
    "Business Cost"
).head()
```

```python
cost_df.to_csv(
    "outputs/threshold_business_costs.csv",
    index=False
)
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    cost_df["Threshold"],
    cost_df["Business Cost"],
    marker="o"
)

plt.title(
    "Business Cost by Decision Threshold"
)

plt.xlabel("Threshold")

plt.ylabel("Estimated Cost (₹)")

plt.tight_layout()

plt.savefig(
    "outputs/business_cost_threshold.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Business Cost Evaluation

Thresholds were evaluated using estimated retention costs:

- False Positive Cost = ₹500
- False Negative Cost = ₹15,000

This analysis helps assess the business impact of threshold selection beyond model performance metrics.

### Threshold Selection

A threshold of **0.45** was selected as it achieved the highest F1-Score (**0.789**), with balanced Precision and Recall (**78.9%**).

### Key Findings

- Best threshold: **0.45**
- F1-Score: **0.789**
- Precision: **78.9%**
- Recall: **78.9%**
- Provides a balanced trade-off between churn detection and retention cost.

# 21. SHAP Analysis

SHAP values were used to explain how individual features influence model predictions.

```python
import shap
import numpy as np

sample_data = X_val.sample(
    n=100,
    random_state=42
)

sample_processed = (
    best_lr.named_steps[
        "preprocessor"
    ].transform(
        sample_data
    )
)

feature_names = (
    best_lr.named_steps[
        "preprocessor"
    ]
    .get_feature_names_out()
)

explainer = shap.LinearExplainer(
    best_lr.named_steps["classifier"],
    sample_processed
)

shap_values = explainer.shap_values(
    sample_processed
)
```

```python
shap_importance = pd.DataFrame({

    "Feature": feature_names,

    "Mean_SHAP": np.abs(
        shap_values
    ).mean(axis=0)

})

shap_importance = (
    shap_importance
    .sort_values(
        by="Mean_SHAP",
        ascending=False
    )
)

shap_importance.head(10)
```

```python
shap.summary_plot(
    shap_values,
    sample_processed,
    feature_names=feature_names,
    show=False
)

plt.savefig(
    "outputs/shap_summary.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Key Findings

- Recency Days was the most influential feature affecting churn predictions.
- Spending behaviour, support interactions, and customer inactivity were strong churn indicators.
- SHAP results closely aligned with the Logistic Regression feature importance analysis.

### Business Interpretation

- Customers with long periods of inactivity are the highest churn-risk group.
- Lower spending and negative support experiences increase churn likelihood.
- Retention efforts should focus on re-engaging inactive customers before churn occurs.

**False Negatives**
- Low predicted risk (0.09–0.34) due to recent activity and higher engagement.
- These customers churned despite appearing healthy in historical data.

### Business Impact

- 31 False Positives resulted in unnecessary retention efforts.
- 31 False Negatives represent missed retention opportunities.
- False Negatives are more costly because they directly impact customer retention and revenue.

# 23. Fairness Assessment

### Objective

Model predictions were evaluated across age groups and city tiers to identify potential prediction disparities across customer segments.

```python
# Create Test Results DataFrame

test_results = X_test.copy()

test_results["actual"] = y_test.values

test_results["predicted"] = y_test_pred

test_results["probability"] = y_test_prob

test_results.head()
```

```python
# Age Group Analysis
fairness_age = (
    test_results
    .groupby("age_group")
    .agg(
        customers=("actual", "count"),
        actual_churn_rate=("actual", "mean"),
        predicted_churn_rate=("predicted", "mean")
    )
    .round(3)
)

fairness_age
```

```python
# City Tier Analysis 
fairness_city = (
    test_results
    .groupby("city_tier")
    .agg(
        customers=("actual", "count"),
        actual_churn_rate=("actual", "mean"),
        predicted_churn_rate=("predicted", "mean")
    )
    .round(3)
)

fairness_city
```

### Fairness Assessment

The model was evaluated across age groups and city tiers to identify potential prediction bias.

- Age-group prediction differences remained below the 15% monitoring threshold.
- Predicted churn rates were broadly aligned with actual churn rates across all city tiers.
- No significant demographic or geographic bias was observed.

### Conclusion

The model demonstrates consistent prediction behavior across the evaluated customer groups. Ongoing fairness monitoring is recommended as part of regular model governance.
